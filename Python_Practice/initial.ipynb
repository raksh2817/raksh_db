```python
## Problem 1 : Two Sums - Given an array of integers nums and an integer target, return the indices of the two numbers such that they add up to target.
## You may assume that each input has exactly one solution, and you may not use the same element twice.



## Solution : 

## Approach 1 : Brute force method to run through nums as i and j pair to match the target o(n^2)
## Approach 2 : using hashmaps to store value for indices and then subrtract num from target and check if remainader is in the dictionary 


def two_sums(nums,target):
    hashmap = {}
    for i, num in enumerate(nums):
        complement = target - num
        if complement in hashmap:
            return [hashmap[complement],i]
        hashmap[num] = i 
    return []

print(two_sums([2, 7, 11, 15], 9))

        
```
